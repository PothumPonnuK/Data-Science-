```python
def get_probabilities(dataset,startrange,endrange):
    from matplotlib import pyplot
    import seaborn as sns
    
    # diagram conversion:
    
    ax=sns.histplot(dataset,kde=True,color='Green')
    pyplot.axvline(startrange,color='Red')
    pyplot.axvline(endrange,color='Red')
    
    #generate a sample:
    
    sample=dataset
    #Calculate the parameter:
    mean=sample.mean()
    std=sample.std()
    print('Mean=%.3f ,Standerd Deviation=% .3f'% (mean,std))
    
    from scipy.stats import norm
    
    distribution=norm(mean,std)
    
    #probability range:

    
    values=[value for value in range (startrange,endrange)]
    probalitities=[distribution.pdf(value) for value in values]
    prob=sum(probalitities)
    print("The area between range ({},{}):{}".format(startrange,endrange,sum(probalitities)))
    return prob
    
    
    
```

```python
 get_probabilities(dataset["hsc_p"],60,70)
```
