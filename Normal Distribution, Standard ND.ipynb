# Standard Normal Distribution

```python
def stdND(dataset):
    mean=dataset.mean()
    std=dataset.std()

    
    #find z_score:
    z_score=[((x-mean)/std) for x in dataset]

    sns.histplot(z_score,kde=True)
    
   
    
    
  
```

```python
stdND(dataset["hsc_p"])
```
